```python
mismatches = []

for record in clean_data:
    for category in [
        "budget_category",
        "mid_range_category",
        "luxury_category"
    ]:
        components = [
            record[category][key]
            for key in [
                "accommodation_range",
                "food_range",
                "activities_range",
                "local_transport_range"
            ]
        ]

        calculated_total = [
            sum(value[0] for value in components),
            sum(value[1] for value in components)
        ]

        if record[category]["total_daily_range"] != calculated_total:
            mismatches.append({
                "id": record["id"],
                "destination": record["destination_name"],
                "category": category,
                "stored_total": record[category]["total_daily_range"],
                "calculated_total": calculated_total
            })

mismatches
```
