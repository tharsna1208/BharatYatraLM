```python
text_columns = [
    "destination_name",
    "state",
    "region",
    "district",
    "trip_types",
    "primary_attractions",
    "activities_available",
    "unique_experiences",
    "hidden_gems",
    "best_seasons",
    "avoid_seasons",
    "ideal_for",
    "ideal_for_why",
    "special_considerations",
    "food_scene",
    "local_culture",
    "festivals_events",
    "local_customs",
    "shopping_highlights",
    "local_cuisine_must_try",
    "user_reviews_summary",
    "sustainability_notes"
]

print("Number of text fields:", len(text_columns))
```

```python
def convert_to_text(value):
    if isinstance(value, list):
        return " ".join(str(item) for item in value)
    elif isinstance(value, dict):
        return " ".join(str(v) for v in value.values())
    else:
        return str(value)
```

```python
for column in text_columns:
    print(column, "→", convert_to_text(df[column].iloc[0])[:100])
```

```python
def create_destination_text(row):
    parts = []

    for column in text_columns:
        value = convert_to_text(row[column])
        parts.append(value)

    return " ".join(parts)


df["destination_text"] = df.apply(create_destination_text, axis=1)

print(df[["destination_name", "destination_text"]].head(1).to_string(index=False))
```

```python
print("Total destinations:", len(df))
print("Missing destination_text:", df["destination_text"].isna().sum())
print("Empty destination_text:", (df["destination_text"].str.strip() == "").sum())
print("Average text length:", df["destination_text"].str.len().mean())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

print("TF-IDF imported successfully")
```

```python
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

tfidf_matrix = tfidf.fit_transform(features_df["destination_text"])

print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity_matrix = cosine_similarity(tfidf_matrix)
print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
goa_index = features_df[
    features_df["destination_name"] == "Goa"
].index[0]

print("Goa index:", goa_index)
```

```python
goa_scores = similarity_matrix[goa_index]

print(goa_scores)
```

```python
similar_destinations = features_df[
    ["destination_name"]
].copy()

similar_destinations["similarity_score"] = goa_scores

similar_destinations = similar_destinations.sort_values(
    by="similarity_score",
    ascending=False
)

print(similar_destinations.head(10).to_string(index=False))
```

```python
top_goa = similar_destinations[
    similar_destinations["destination_name"] != "Goa"
].drop_duplicates(
    subset="destination_name"
).head(5)

print(top_goa.to_string(index=False))
```

```python
def recommend_similar_destinations(destination_name, top_n=5):
    matches = features_df[
        features_df["destination_name"] == destination_name
    ]

    if matches.empty:
        return f"Destination '{destination_name}' not found."

    index = matches.index[0]

    scores = similarity_matrix[index]

    results = features_df[["destination_name"]].copy()
    results["similarity_score"] = scores

    results = results[
        results["destination_name"] != destination_name
    ]

    results = results.drop_duplicates(
        subset="destination_name"
    )

    return results.sort_values(
        by="similarity_score",
        ascending=False
    ).head(top_n)
```

```python
print(recommend_similar_destinations("Goa"))
```

```python
print(recommend_similar_destinations("Ladakh (Leh)"))
```

```python
import pickle

with open("recommendation_data.pkl", "wb") as f:
    pickle.dump(
        {
            "tfidf": tfidf,
            "tfidf_matrix": tfidf_matrix,
            "similarity_matrix": similarity_matrix,
            "features_df": features_df
        },
        f
    )

print("Recommendation data saved successfully")
```
